```python
dfa = init_automaton_ex_1_2()
print(f"Start: {dfa._start_state}")
print(f"Func count: {len(dfa._transitions)}")
print(f"Final: {dfa._accept_states}")

test_words = ["ab", "aab", "bab", "a", "b", "", "abab", "abbbbbbb"]
for word in test_words:
    result = dfa.check(word, verbose=False)
    expected = word.endswith("ab")
    status = "" if result == expected else "Error: "
    print(f"  {status} '{word}': {result} (expected {expected})")
```

```python
min_dfa = minimize_dfa(dfa)
print(f"Start: {min_dfa._start_state}")
print(f"Func count: {len(min_dfa._transitions)}")
print(f"Final: {min_dfa._accept_states}")

for word in test_words:
    result = min_dfa.check(word, verbose=False)
    expected = word.endswith("ab")
    status = "" if result == expected else "Error: "
    print(f"  {status} '{word}': {result} (expected {expected})")
```

```python
nfa = init_nfa_ex_1_2()
print(f"Start: {nfa._start_state}")
print(f"Func count: {len(nfa._transitions)}")
print(f"Final: {nfa._accept_states}")

test_words_nfa = ["ab", "aab", "bab", "a", "b", "", "abab", "ba", "abbbbbbbb"]
for word in test_words_nfa:
    result = nfa.check(word, verbose=False)
    expected = "ab" in word
    status = "" if result == expected else "Error: "
    print(f"  {status} '{word}': {result} (expected {expected})")
```

```python
dfa_from_nfa = nfa_to_dfa(nfa)
print(f"Start: {dfa_from_nfa._start_state}")
print(f"Func count: {len(dfa_from_nfa._transitions)}")
print(f"Final: {dfa_from_nfa._accept_states}")

for word in test_words_nfa:
    result = dfa_from_nfa.check(word, verbose=False)
    expected = "ab" in word
    status = "" if result == expected else "Error: "
    print(f"  {status} '{word}': {result} (expected {expected})")
```

```python
min_dfa_from_nfa = minimize_dfa(dfa_from_nfa)
print(f"Start: {min_dfa_from_nfa._start_state}")
print(f"Func count: {len(min_dfa_from_nfa._transitions)}")
print(f"Final: {min_dfa_from_nfa._accept_states}")

for word in test_words_nfa:
    result = min_dfa_from_nfa.check(word, verbose=False)
    expected = "ab" in word
    status = "" if result == expected else "Error: "
    print(f"  {status} '{word}': {result} (expected {expected})")
```
